# Combine new and old csv

```python
main_df = pd.read_csv("../data/test/full_test.csv")
main_df
```

```python
SEED=0
NUM_EXAMPLES_PER_LABEL=2

# MODEL="gpt-3.5-turbo"
MODEL="gpt-4"
PROMPT_TYPE="depression"

df = pd.read_csv(f"../results/few_shot/few_shot_{PROMPT_TYPE}_{MODEL}_num_examples_{NUM_EXAMPLES_PER_LABEL}_seed_{SEED}_old.csv")
print(df.shape)
df = df.dropna().reset_index(drop=True)
df
```

```python
df_new = pd.read_csv(f"../results/few_shot/few_shot_{PROMPT_TYPE}_{MODEL}_num_examples_{NUM_EXAMPLES_PER_LABEL}_seed_{SEED}_new.csv")
df_new.shape
```

```python
result_dict = dict()
for i in range(len(df)):
    key = df.at[i, f'id']
    val = df.at[i, f'results_{PROMPT_TYPE}_{MODEL}']
    result_dict[key] = val
    
for i in range(len(df_new)):
    key = df_new.at[i, f'id']
    val = df_new.at[i, f'results_{PROMPT_TYPE}_{MODEL}']
    result_dict[key] = val
    
len(result_dict)
```

```python
main_df[f'results_{PROMPT_TYPE}_{MODEL}'] = main_df['id'].apply(lambda x: result_dict[x])
main_df
```

```python
main_df.to_csv(f"../results/few_shot/few_shot_{PROMPT_TYPE}_{MODEL}_num_examples_{NUM_EXAMPLES_PER_LABEL}_seed_{SEED}.csv", index=False)
```
