```python
df = pd.read_csv("../results/zero_shot/zero_shot_depression_anxiety_comorbidity_gpt-3.5-turbo_seed_0.csv")
df
```

```python
results = df['results_depression_anxiety_comorbidity_gpt-3.5-turbo']
len(results)
```

```python
depression = 0
anxiety = 0
invalid = 0
total = 0
for result in tqdm(results):
    try:
        result = ast.literal_eval(result)
        if result['depression'].lower() == "yes":
            depression += 1

        if result['anxiety'].lower() == "yes":
            anxiety += 1

        total += 1
    
    except:
        invalid += 1

depression, (depression/total), anxiety, (anxiety/total)
```
