```python
def get_results(i):
    preds = pd.read_csv(file_path + files[i])['results' + files[i][9:-11]].tolist()
    pd.read_csv(file_path + files[i])
    labels = pd.read_csv(file_path + files[i])['labels'].tolist()
    a_true = [int(x[4]) for x in labels]
    d_true = [int(x[1]) for x in labels]
    return preds, a_true, d_true
```
