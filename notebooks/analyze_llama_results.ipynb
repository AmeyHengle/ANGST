```python
import pandas as pd
import numpy as np
from sklearn.metrics import precision_recall_fscore_support
```

```python
def generate_label(preds_str):
    if "yes" in preds_str.lower():
        return 1
    elif "no" in preds_str.lower():
        return 0
    else:
     return -1
```

# Zero Shot

```python
seed=1
# data_type = "depression"
data_type = "anxiety"

model = "llama_chat_7b"
# model = "llama_chat_13b"

df = pd.read_csv(f"../results/zero_shot/zero_shot_{data_type}_llama_{model}_seed_{seed}_v{seed}.csv")
df['preds'] = [generate_label(str(pred).replace("\n", "")) for pred in df[f"results_{data_type}_llama_{model}"]]
print(df["preds"].value_counts())
```

```python
df = pd.read_csv(f"../results/zero_shot/zero_shot_{data_type}_{model}_seed_{seed}_v{seed}.csv")
df[f"results_{data_type}_{model}"] = df[f"results_{data_type}_{model}"].apply(lambda x: str(x).replace("\n", ""))
df['preds'] = [generate_label(pred) for pred in df[f"results_{data_type}_{model}"]]
print(df["preds"].value_counts())
```

```python
precision, recall, fscore, support = precision_recall_fscore_support(y_true=df['depression_label'].tolist(), y_pred=df['preds'].tolist(), average="macro")
results = {
    "precision": precision,
    "recall": recall,
    "fscore": fscore,
}
results
```
